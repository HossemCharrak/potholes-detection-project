## converting the torch model to onnx model to use it in the web app

```python
import torch
import sys
from ultralytics import YOLO

def convert_to_onnx(pt_path: str, onnx_path: str):
    # Load the YOLOv8 model
    model = YOLO(pt_path)
    
    # Export the model to ONNX
    model.export(format='onnx', 
                opset=12,  # ONNX opset version
                simplify=True,  # Simplify the model
                dynamic=True)  # Enable dynamic axes
    
    print(f"Model converted and saved to {onnx_path}")

if __name__ == "__main__":
    pt_path = "/kaggle/working/runs/detect/yolov8s_custom/weights/best.pt"  # Your PyTorch model path
    onnx_path = "/kaggle/working/runs/detect/yolov8s_custom/weights/best.onnx"  # Output ONNX model path
    convert_to_onnx(pt_path, onnx_path)
```

```python
import onnxruntime as ort
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ---- Settings ----
onnx_model_path = "/kaggle/working/runs/detect/yolov8s_custom/weights/best.onnx"
image_path = "/kaggle/input/pathole-dataset-665-img/test/images/img-337_jpg.rf.aee9516f4db8d93d1fcf7071196974f9.jpg"
input_size = 640
confidence_threshold = 0.4
iou_threshold = 0.5

# ---- Load image ----
image = cv2.imread(image_path)
original_image = image.copy()
image_resized = cv2.resize(image, (input_size, input_size))
image_rgb = cv2.cvtColor(image_resized, cv2.COLOR_BGR2RGB)
input_tensor = image_rgb.transpose(2, 0, 1).astype(np.float32) / 255.0
input_tensor = np.expand_dims(input_tensor, axis=0)

# ---- Load ONNX model ----
session = ort.InferenceSession(onnx_model_path)
inputs = {session.get_inputs()[0].name: input_tensor}
outputs = session.run(None, inputs)

# ---- Postprocess output ----
output = outputs[0][0]  # Shape: (5, 8400)
output = output.transpose(1, 0)  # Shape: (8400, 5)
boxes = output[:, :4]
scores = output[:, 4]

# Filter by confidence
indices = np.where(scores > confidence_threshold)[0]
boxes = boxes[indices]
scores = scores[indices]

# Convert xywh → xyxy
def xywh2xyxy(boxes):
    converted = np.copy(boxes)
    converted[:, 0] = boxes[:, 0] - boxes[:, 2] / 2  # x1
    converted[:, 1] = boxes[:, 1] - boxes[:, 3] / 2  # y1
    converted[:, 2] = boxes[:, 0] + boxes[:, 2] / 2  # x2
    converted[:, 3] = boxes[:, 1] + boxes[:, 3] / 2  # y2
    return converted

boxes = xywh2xyxy(boxes)

# Rescale boxes to original image size
h0, w0 = original_image.shape[:2]
scale_x = w0 / input_size
scale_y = h0 / input_size
boxes[:, [0, 2]] *= scale_x
boxes[:, [1, 3]] *= scale_y

# Apply NMS
indices = cv2.dnn.NMSBoxes(
    bboxes=boxes.tolist(),
    scores=scores.tolist(),
    score_threshold=confidence_threshold,
    nms_threshold=iou_threshold
)

# Draw boxes
for i in indices:
    i = i[0] if isinstance(i, (list, tuple, np.ndarray)) else i
    x1, y1, x2, y2 = boxes[i].astype(int)
    cv2.rectangle(original_image, (x1, y1), (x2, y2), (0, 255, 0), 2)
    label = f"Pothole {scores[i]:.2f}"
    cv2.putText(original_image, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 1)

# ---- Show final image ----
plt.figure(figsize=(10, 8))
plt.imshow(cv2.cvtColor(original_image, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title("Pothole Detection (ONNX)")
plt.show()
```

```python
import onnxruntime as ort
import cv2
import numpy as np

# ---- Settings ----
onnx_model_path = "/kaggle/working/runs/detect/yolov8s_custom/weights/best.onnx"
video_path = "/kaggle/input/potholes-detection-inference-on-videos/pothole.mp4"
output_path = "output_pothole_detection_onnx.mp4"
input_size = 640
confidence_threshold = 0.4
iou_threshold = 0.5
max_frames = 300  # Set a frame limit if needed

# ---- Load ONNX model ----
session = ort.InferenceSession(onnx_model_path)
input_name = session.get_inputs()[0].name

# ---- Video setup ----
cap = cv2.VideoCapture(video_path)
fps = cap.get(cv2.CAP_PROP_FPS)
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
output_size = (frame_width, frame_height)

fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_path, fourcc, fps, output_size)

# ---- Helper functions ----
def preprocess(frame):
    resized = cv2.resize(frame, (input_size, input_size))
    rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)
    input_tensor = rgb.transpose(2, 0, 1).astype(np.float32) / 255.0
    return np.expand_dims(input_tensor, axis=0)

def xywh2xyxy(boxes):
    converted = np.copy(boxes)
    converted[:, 0] = boxes[:, 0] - boxes[:, 2] / 2  # x1
    converted[:, 1] = boxes[:, 1] - boxes[:, 3] / 2  # y1
    converted[:, 2] = boxes[:, 0] + boxes[:, 2] / 2  # x2
    converted[:, 3] = boxes[:, 1] + boxes[:, 3] / 2  # y2
    return converted

# ---- Inference loop ----
frame_count = 0
while cap.isOpened() and frame_count < max_frames:
    ret, frame = cap.read()
    if not ret:
        break

    input_tensor = preprocess(frame)
    outputs = session.run(None, {input_name: input_tensor})
    output = outputs[0][0].transpose(1, 0)  # (8400, 5)

    boxes = output[:, :4]
    scores = output[:, 4]
    indices = np.where(scores > confidence_threshold)[0]
    boxes = boxes[indices]
    scores = scores[indices]

    if boxes.shape[0] == 0:
        out.write(frame)
        frame_count += 1
        continue

    boxes = xywh2xyxy(boxes)

    scale_x = frame.shape[1] / input_size
    scale_y = frame.shape[0] / input_size
    boxes[:, [0, 2]] *= scale_x
    boxes[:, [1, 3]] *= scale_y

    nms_indices = cv2.dnn.NMSBoxes(
        bboxes=boxes.tolist(),
        scores=scores.tolist(),
        score_threshold=confidence_threshold,
        nms_threshold=iou_threshold
    )

    for i in nms_indices:
        i = i[0] if isinstance(i, (list, tuple, np.ndarray)) else i
        x1, y1, x2, y2 = boxes[i].astype(int)
        cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
        label = f"Pothole {scores[i]:.2f}"
        cv2.putText(frame, label, (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 1)

    out.write(frame)
    frame_count += 1

# ---- Cleanup ----
cap.release()
out.release()
print("Video saved to:", output_path)
```
