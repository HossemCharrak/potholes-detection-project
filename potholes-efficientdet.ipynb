```python
import pandas as pd
def coco_to_csv(json_data, image_dir):
    id_to_filename = {img['id']: img['file_name'] for img in json_data['images']}
    id_to_class = {cat['id']: cat['name'] for cat in json_data['categories']}

    rows = []
    for ann in json_data['annotations']:
        image_id = ann['image_id']
        file_name = id_to_filename.get(image_id)
        if not file_name:
            continue  # skip if image_id not found

        image_path = os.path.join(image_dir, file_name)
        if not os.path.exists(image_path):
            # Sometimes file_name might be wrong or file might be missing
            print(f"Warning: Image not found: {image_path}")
            continue

        bbox = ann['bbox']
        x_min, y_min, width, height = bbox
        x_max = x_min + width
        y_max = y_min + height

        category_id = ann['category_id']
        class_name = id_to_class.get(category_id, "undefined")
        rows.append([image_path, x_min, y_min, x_max, y_max, class_name])

    df = pd.DataFrame(rows, columns=["image_path", "xmin", "ymin", "xmax", "ymax", "class_name"])
    return df

```

```python
# Convert
train_df = coco_to_csv(train_data, train_image_dir).sample(frac=1, random_state=42).reset_index(drop=True)
valid_df = coco_to_csv(valid_data, valid_image_dir).sample(frac=1, random_state=42).reset_index(drop=True)
test_df = coco_to_csv(test_data, test_image_dir).sample(frac=1, random_state=42).reset_index(drop=True)
```
